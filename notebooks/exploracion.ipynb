# Exploración — INFONAVIT vs Rezago habitacional

Validación rápida de los datasets generados por el pipeline y obtención de los insights principales.

```python
import pandas as pd

pd.options.display.float_format = "{:,.2f}".format

fct = pd.read_csv("../data/processed/fct_creditos_anual.csv", dtype={"clave_estado": str})
rezago = pd.read_csv("../data/processed/dim_rezago.csv", dtype={"clave_estado": str})
resumen = pd.read_csv("../data/processed/resumen_ejecutivo.csv", dtype={"clave_estado": str})

print(f"Créditos: {len(fct):,} filas, {fct['anio'].min()}-{fct['anio'].max()}")
print(f"Rezago:   {len(rezago)} estados")
print(f"Resumen:  {len(resumen)} estados con métricas ejecutivas")
```

## Totales nacionales

```python
print(f"Créditos totales:   {fct['acciones'].sum():,.0f}")
print(f"Monto acumulado:    ${fct['monto'].sum()/1e12:.2f} billones MXN")
print(f"Monto promedio:     ${fct['monto'].sum()/fct['acciones'].sum():,.0f} MXN")
```

## Top estados por volumen

```python
resumen.head(10)[["estado", "region", "creditos_totales", "cagr"]]
```

## Hallazgo principal: brecha de cobertura

El ratio de cobertura vs rezago muestra diferencias de más de 100x entre estados.

```python
print("Mejor cobertura:")
print(resumen.nlargest(5, "ratio_cobertura_vs_rezago")[
    ["estado", "creditos_totales", "viviendas_en_rezago", "ratio_cobertura_vs_rezago"]
].to_string(index=False))

print("\nPeor cobertura:")
print(resumen.nsmallest(5, "ratio_cobertura_vs_rezago")[
    ["estado", "creditos_totales", "viviendas_en_rezago", "ratio_cobertura_vs_rezago"]
].to_string(index=False))
```

## Crecimiento anual (CAGR 2015-2024)

```python
print("Mayor crecimiento:")
print(resumen.nlargest(5, "cagr")[["estado", "cagr", "creditos_ultimo_anio"]].to_string(index=False))

print("\nMayor contracción:")
print(resumen.nsmallest(5, "cagr")[["estado", "cagr", "creditos_ultimo_anio"]].to_string(index=False))
```
